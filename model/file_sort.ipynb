```python
import pandas as pd
import csv
import os
import shutil
from pathlib import Path
```

```python
train_dir = Path('sorted/train')
classes_file = 'train/_classes.csv'
img_dir = Path('train')
```

```python
val_dir = Path('sorted/valid')
classes_file = 'valid/_classes.csv'
img_dir = Path('valid')
```

```python
for i in range(1, 8):
    class_dir = val_dir / f'class_{i}'
    os.makedirs(class_dir, exist_ok=True)
```

```python
with open(classes_file, 'r') as f:
    reader = csv.reader(f)
    next(reader)

    for row in reader:
        if len(row) < 8:
            print('Skipping invalid row: {row}')
            continue

        filename = row[0].strip()
        class_values = [int(val) for val in row[1:8]]

        try:
            class_idx = class_values.index(1) + 1
            src_path = img_dir / filename
            dst_path = val_dir / f'class_{class_idx}' / filename

            if os.path.exists(src_path):
                shutil.copy2(src_path, dst_path)
                print(f'Moved {filename} to class_{class_idx}')
            else:
                print(f'File not found: {src_path}')
        
        except ValueError:
            print(f'No class found for {filename}, class values: {class_values} ')

print('Files copied')
```
